The table should:

* Show the size of vulnerable populations in Argentina, Malawi, Nepal, and the United Kingdom for the years 1982, 1987, and 1992
* Include a summary column showing the mean average size of vulnerable populations across these three years
* Include a spanner heading above the 1982, 1987, and 1992 columns called ‘Vulnerable population by year’
* Include a stubhead label heading above the countries column called 'Developer'
* Have commas to separate thousands
* Be formatted so that:
  * The spanner heading is a dark green with a HEX code of #324027 and the text is white
  * The stubhead label heading is a dark green with a HEX code of #324027 and the text is white and bold
  * The column labels are an earthy green with a HEX code of #48553F and the text is white



```python

# Imports
import os
import pandas as pd
import polars as pl
from great_tables import GT, vals, loc, style


# Set working directory to the learning_resources folder
repo_root = os.path.abspath(
    os.path.join(os.path.dirname(os.path.abspath("making_tables_exercise.ipynb")), ("../../../../.."))
)

print("repo_root:", repo_root)

learning_resources_dir = os.path.join(repo_root, "learning_resources")
os.chdir(learning_resources_dir)
print("Current working directory:", os.getcwd())

# Load data
vulnerable_path = os.path.join(learning_resources_dir, "data", "vulnerable.csv")
vuln = pd.read_csv(vulnerable_path)
```

```python
# Get table population in 1982, 1987 and 1992 for Argentina, Malawi, Nepal and UK

vuln_countries_year = vuln.loc[
    vuln["country"].isin(["Argentina", "Malawi", "Nepal", "United Kingdom"])
    & vuln["year"].isin([1982, 1987, 1992]),
    ["country", "year", "vulnerable_pop"],
]

# Reshaping the data

vuln_countries_year = vuln_countries_year.pivot_table(
    values="vulnerable_pop",
    index="country",
    columns="year"
)

vuln_countries_year
```

```python
# Add summary column for average of the three years

vuln_countries_year_average = vuln_countries_year.copy()

vuln_countries_year_average["Average"] = vuln_countries_year.mean(axis=1).round(1)

vuln_countries_year_average
```

```python
# Make the dataframe into a Great Table object
# Reset the index to make country a column instead of an index

vuln_countries_year_GT = (GT(vuln_countries_year_average.reset_index(), rowname_col="country")

# Add a spanner heading to the table
.tab_spanner(
    label="Vulnerable Population by Year",
    columns=["1982", "1987", "1992"]
)

# Add a stub heading to the table
.tab_stubhead(label="Developer")
)


vuln_countries_year_GT           
```

```python
# Add commas to seperate thousands 

vuln_countries_year_GT = vuln_countries_year_GT.fmt_number(
    columns = ["1982", "1987", "1992", "Average"],
    decimals = 0,
    use_seps = True)

vuln_countries_year_GT
```
